```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib.patches import Arc

# -----------------------
# Parametri (invariati rispetto alla versione verticale)
# -----------------------
b = 1.3
alpha_deg = 20
alpha = np.deg2rad(alpha_deg)

fig, ax = plt.subplots(figsize=(6,3.2))   # <-- figsize invertita: larga, non alta

z_top, z_bot = 5.5, -3.0

# ==========================================================
# Asintoto d'INGRESSO: era verticale (x=0), ora ORIZZONTALE (X=0)
# Zh = -z: il punto piu' lontano (z_top) diventa il piu' a sinistra
# ==========================================================
ax.plot([-z_top,-z_bot],[0,0], '--', color='k', lw=1.1)

# ==========================================================
# Asintoto di USCITA: pendenza tan(alpha), ora verso il basso a destra
# ==========================================================
Zh_out = np.array([0, -z_bot])
x_out = -np.tan(alpha) * Zh_out
ax.plot(Zh_out, x_out, '--', color='k', lw=1.1)

# ==========================================================
# Lente M e punto b: ora il segmento b e' VERTICALE (giusto, e' un
# parametro d'impatto perpendicolare alla linea di vista)
# ==========================================================
ax.plot(0, -b, 'ko', ms=6)
ax.text(-0.15, -b-0.35, "M", fontsize=13, ha='center')

ax.plot([0,0],[-b,0], ':', color='k', lw=1)
ax.plot(0, 0, 'ko', ms=6)
ax.text(0.22, -b/2, "b", fontsize=12, style='italic')

# ==========================================================
# Raggio deflesso
# ==========================================================
z = np.linspace(z_top, z_bot, 500)
k = 1.1
w = 1/(1+np.exp(k*z))
dz = z[0]-z[1]
x = -np.tan(alpha) * np.cumsum(w)*dz

ax.plot(-z, x, color='k', lw=1.8)   # <-- Zh = -z al posto di x, x al posto di z

# ==========================================================
# Delta z: ora e' un intervallo ORIZZONTALE (era verticale)
# ==========================================================
dz_half = 1.3
ax.annotate("", xy=(-dz_half,0.10), xytext=(0,0.10),
            arrowprops=dict(arrowstyle="<->", lw=1))
ax.text(-dz_half*0.55, 0.28, r"$\Delta z$", fontsize=12, ha='center')

# ==========================================================
# Angolo di deflessione: vertice in b=(0,0), tra le due tratteggiate
# ==========================================================
arc = Arc((0,0), width=3, height=3,
          theta1=-alpha_deg, theta2=0, lw=1.3, color='k')
ax.add_patch(arc)
mid_ang = np.deg2rad(-alpha_deg/2)
ax.text(0.85*np.cos(mid_ang)+0.70, 0.85*np.sin(mid_ang)-0.25,
        r"$\hat{\alpha}$", fontsize=13)

# ==========================================================
ax.set_xlim(-z_top-0.3, -z_bot+0.3)
ax.set_ylim(-2.6, 1.4)
ax.set_aspect('equal')
ax.axis('off')
plt.tight_layout()
plt.savefig("deflection_horizontal.pdf", bbox_inches="tight")
plt.savefig("deflection_horizontal.png", dpi=600, bbox_inches="tight")
print("ok")
```
